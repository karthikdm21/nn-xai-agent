```python
import sys
sys.path.append('..')
import torch
from torchvision import transforms
from PIL import Image
import glob
import base64
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import io

from tools.shap_tool import explain_shap

print("Imports done ✅")
```

```python
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
])

image_files = glob.glob("../data/HAM10000_images_part_1/*.jpg")[:11]
tensors = [transform(Image.open(f).convert("RGB")) for f in image_files]

background = torch.stack(tensors[:10])
test_image = tensors[10].unsqueeze(0)

print(f"Loaded {len(image_files)} images")
print(f"Test image: {image_files[10]}")
```

```python
print("Running SHAP — wait 30-60 seconds...")
result = explain_shap(test_image, background, pred_class=0)
print("Done!")
print(f"Status:         {result['status']}")
print(f"Mean SHAP:      {result['mean_shap']}")
print(f"Max SHAP:       {result['max_shap']}")
print(f"Interpretation: {result['interpretation']}")
```

```python
if result['plot_b64']:
    img_data = base64.b64decode(result['plot_b64'])
    img      = Image.open(io.BytesIO(img_data))
    plt.figure(figsize=(12, 5))
    plt.imshow(img)
    plt.axis('off')
    plt.title("SHAP explanation — brighter areas = more important to prediction")
    plt.show()
    print("✅ SHAP plot displayed — this is what the agent will show users")
else:
    print("No plot generated:", result.get('error'))
```
